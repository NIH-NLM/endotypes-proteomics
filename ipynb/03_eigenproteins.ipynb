# Step 03. Eigenproteins: one number per patient per module

A module's eigenprotein (WGCNA calls it an eigengene) is the first principal component of its
proteins: a single value per patient summarising the whole module. 7,288 proteins become 52 numbers,
and those 52 are the space patients get clustered in.

```r
source("../src/paths.R")
suppressMessages(library(WGCNA))
w <- readRDS(art("wgcna_A.rds"))
ME <- w$ME
dim(ME)
```

## `MEgrey` is dropped

Grey is the unassigned bin, not a module. Its eigenprotein averages over 1,107 reagents that
share nothing. Left in, it topped the module–trait table, a meaningless result that looked like a
finding.

```r
ME <- ME[, colnames(ME) != "MEgrey", drop = FALSE]
ncol(ME)
```

## Eigenprotein scale

`moduleEigengenes` returns the first principal component (PC1) scaled to unit norm, not unit variance. Every entry is therefore
on the order of `1/sqrt(n)` ≈ 0.10, and any group mean computed on the raw eigenproteins comes out
around ±0.1 and looks flat when it is not.

Standardise before reading group means. Then the numbers are in standard deviation (SD) units and mean what they appear
to mean.

```r
S <- scale(ME)
rbind(raw = round(apply(ME, 2, sd)[1:4], 4),
      scaled = round(apply(S, 2, sd)[1:4], 4))
```

```r
saveRDS(list(ME = ME, S = S), art("eigengenes_A.rds"))
```
